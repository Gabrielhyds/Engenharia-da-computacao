```python
#1- Crie um programa que solicite ao usuário uma temperatura em graus Celsius e converta
#para Fahrenheit.

#Entrada:
    ##• Um número representando a temperatura em Celsius.
#Saída:
    ##• A temperatura equivalente em Fahrenheit.
# Fórmula: F = (9/5) * C + 32

celsius = int(input("Digite uma temperatura em celsius:"))

fahrenheit = (9/5) * celsius + 32

print(f"A temperatura equivalente em Fahrenheit eh: {fahrenheit}")
```

```python
#2 - Escreva um programa que peça ao usuário para digitar uma frase e conte quantas vogais (a,
#e, i, o, u) existem na frase, ignorando maiúsculas e minúsculas.

##Entrada:
    #• Uma string representando a frase.
##Saída:
    #• Um número indicando a quantidade de vogais.

frase = str(input("Digite uma frase:"))

a = frase.count("a")
e = frase.count("e")
i = frase.count("i")
o = frase.count("o")
u = frase.count("u")

print(f"Essa frase tem: {a+e+i+o+u} vogais")

```

```python
#3 - Crie uma função que recebe um número inteiro e retorna True se ele for primo, e False caso contrário.

##Entrada:
  #• Um número inteiro.
##Saída:
  #• Um valor booleano indicando se o número é primo ou não.

def primo(numero):
    if numero < 2:
        return False
    for i in range(2, numero):
        if numero % i == 0:
            return False
    return True

print(primo(6))
```

```python
#4 - Implemente o algoritmo de ordenação Bubble Sort sem utilizar funções prontas como sort() ou sorted().

lista = [5, 2, 9, 1, 5, 6]

for i in range(len(lista)):
    for j in range(len(lista) - 1):
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]

print(lista)
```
