```python
#Exercício 2 — Calculadora simples
#Escreva um programa que leia dois números e um operador (+, -, *, /) digitados pelo usuário e exiba
#o resultado da operação correspondente.

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
operador = input("Digite o operador (+, -, *, /): ")
if(operador == "+"):
    resultado = num1 + num2
elif(operador == "-"):
    resultado = num1 - num2
elif(operador == "*"):
    resultado = num1 * num2
elif(operador == "/"):
    resultado = num1 / num2
else:
    print("Operador inválido")
print(f"O resultado da operação é: {resultado}")
```

```python
#Exercício 3 — Média de três notas
#Escreva um programa que leia três notas de um aluno, calcule a média aritmética e informe se o
#aluno foi Aprovado (média ≥ 7), está de Recuperação (média entre 5 e 6.9) ou foi Reprovado
#(média < 5).

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))
media = (nota1 + nota2 + nota3) / 3
if(media >= 7):
    print("Aprovado")
elif(media >= 5 and media <= 6.9):
    print("Recuperação")
else:
    print("Reprovado")
```

```python
#Exercício 4 — Par ou ímpar
#Escreva uma função par_ou_impar(n) que receba um número inteiro e retorne a string "par" ou
#"ímpar", conforme o caso.

def par_ou_impar(n):
    if(n % 2 == 0):
        return "par"
    else:
        return "ímpar"

print(par_ou_impar(11))
```

```python
#Exercício 5 — Soma dos elementos de uma lista
#Dada uma lista de números inteiros, escreva um programa que calcule e imprima a soma de todos
#os seus elementos, sem usar a função embutida sum().

lista = [1, 2, 3, 4, 5]
soma = 0
for i in lista:
    soma += i
print(soma)
```

```python
#Exercício 6 — Maior e menor valor
#Escreva um programa que receba uma lista de números (via input do usuário, separados por
#espaço) e determine, sem usar as funções max() e min(), qual é o maior e o menor valor da lista.

lista = input("Digite os números da lista, separados por espaço: ").split()
lista = [int(i) for i in lista]
maior = lista[0]
menor = lista[0]
for i in lista:
    if(i > maior):
        maior = i
    if(i < menor):
        menor = i
print(f"O maior valor da lista é: {maior}")
print(f"O menor valor da lista é: {menor}")
```

```python
#Exercício 7 — Contagem de palavras
#Escreva um programa que receba uma frase digitada pelo usuário e conte quantas vezes cada
#palavra aparece, armazenando o resultado em um dicionário e imprimindo-o ao final.

frase = input("Digite uma frase: ").split()
contagem = {}
for i in frase:
    if(i in contagem):
        contagem[i] += 1
    else:
        contagem[i] = 1
print(contagem)
```

```python
#Exercício 8 — Fatorial recursivo
#Implemente uma função recursiva fatorial(n) que calcule o fatorial de um número inteiro não
#negativo n.

def fatorial(n):
    if(n == 0):
        return 1
    else:
        return n * fatorial(n - 1)

print(fatorial(5))
```

```python
#Exercício 9 — Operações com vetores
#Representando vetores como listas de números, implemente três funções: soma_vetores(v1,
#v2), que retorna a soma elemento a elemento; produto_escalar(v1, v2), que retorna o
#produto escalar entre dois vetores; e norma(v), que retorna a norma (magnitude) do vetor.

def soma_vetores(v1, v2):
    if(len(v1) != len(v2)):
        return "Vetores de tamanhos diferentes"

    soma = []
    for i in range(len(v1)):
        soma.append(v1[i] + v2[i])
    return soma

print(soma_vetores([1, 2, 3], [4, 5, 6]))
```

```python
#Exercício 10 — Verificação de palíndromo
#Escreva uma função eh_palindromo(texto) que verifique se uma string é um palíndromo,
#ignorando espaços, pontuação e diferenças entre maiúsculas e minúsculas (ex.: "Ame o poema"
#deve retornar True).

def eh_palindromo(texto):
  texto = texto.lower()
  texto = texto.replace(" ", "")
  texto = texto.replace(".", "")
  texto = texto.replace(",", "")
  texto = texto.replace("!", "")
  texto = texto.replace("?", "")
  if(texto == texto[::-1]):
    return True
  else:
    return False

print(eh_palindromo("Ame o poema"))
```

```python
#Exercício 11 — Busca binária
#Implemente o algoritmo de busca binária em uma função busca_binaria(lista, alvo), que
#recebe uma lista ordenada de números e um valor a ser procurado, retornando o índice do elemento
#encontrado ou -1 caso não exista

def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1
    while(esquerda <= direita):
        meio = (esquerda + direita) // 2
        if(lista[meio] == alvo):
            return meio
        elif(lista[meio] < alvo):
            esquerda = meio + 1
        else:
            direita = meio - 1
    return -1

print(busca_binaria([1, 2, 3, 4, 5], 3))
```

```python
#Exercício 12 — Ordenação (Bubble Sort)
#Implemente o algoritmo de ordenação Bubble Sort em uma função bubble_sort(lista), que
#ordene uma lista de números em ordem crescente, sem utilizar a função sorted() ou o método
#.sort().

def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n-i-1):
            if(lista[j] > lista[j+1]):
                lista[j], lista[j+1] = lista[j+1], lista[j]
    return lista

print(bubble_sort([5, 3, 2, 1, 4]))
```

```python
#Exercício 13 — Multiplicação de matrizes
#Representando matrizes como listas de listas, implemente uma função
#multiplicar_matrizes(a, b) que retorne o produto matricial entre as matrizes a e b, validando
#se as dimensões são compatíveis.

def multiplicar_matrizes(a, b):
    if(len(a[0]) != len(b)):
        return "As matrizes não podem ser multiplicadas"
    for i in range(len(a)):
        for j in range(len(b[0])):
            for k in range(len(b)):
                a[i][j] = a[i][j] + a[i][k] * b[k][j]
    return a

print(multiplicar_matrizes([[1, 2], [3, 4]], [[5, 6], [7, 8]]))
```

```python
#Exercício 14 — Crivo de Eratóstenes
#Implemente o algoritmo do Crivo de Eratóstenes em uma função crivo_eratostenes(n) que
#retorne a lista de todos os números primos menores ou iguais a n.

def crivo_eratostenes(n):
    primos = [True] * (n + 1)
    primos[0] = False
    primos[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if(primos[i]):
            for j in range(i * i, n + 1, i):
                primos[j] = False
    primos = [i for i in range(n + 1) if primos[i]]
    return primos

print(crivo_eratostenes(20))
```
